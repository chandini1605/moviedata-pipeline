```python
#Importing libraries
import pandas as pd
from pathlib import Path
```

```python

#Set the project directory
project_dir = Path(r"C:\Users\Admin\Desktop\movie-pipeline")

# Locate the dataset files
movies_path = project_dir / "data" / "movies.csv"
ratings_path = project_dir / "data" / "ratings.csv"

# Check whether the dataset files exist
if not movies_path.exists() or not ratings_path.exists():
    raise FileNotFoundError(
        "Dataset files not found. Please check the project path and data folder."
    )
```

```python
# Read the datasets
movies = pd.read_csv(movies_path)
ratings = pd.read_csv(ratings_path)

print("Datasets loaded successfully!")
```

```python
#Displaying Datasets
print(movies.head())
print("\n")
print(ratings.head())
```

```python
#Dataset dimensions
print("Movies:", movies.shape)
print("Ratings:", ratings.shape)
```

```python
#Missing values
print("Movies\n")
print(movies.isnull().sum())
print("\n")
print("ratings\n")
print(ratings.isnull().sum())
```

```python
#Duplicate records
print("Movies duplicates:", movies.duplicated().sum())
print("Movies duplicates with primary key(movieid):", movies.duplicated(subset="movieId").sum())
print("Ratings duplicates:", ratings.duplicated().sum())
print("Ratings duplicates with composite key(movieid,userid):", ratings.duplicated(subset=["userId", "movieId"]).sum())
```

```python
print(movies.describe())
print("\n")
print(ratings["rating"].describe())
```

```python
print("movies")
print(movies.dtypes)
print("\n")
print("ratings")
print(ratings.dtypes)
```

```python
# Rating distribution
print("Rating counts:")
print(ratings["rating"].value_counts().sort_index())
```

```python
# Number of unique users and movies
print("Total unique users:", ratings["userId"].nunique())
print("Total unique movies rated:", ratings["movieId"].nunique())
```

```python
# Genre Analysis
# Check the number of movies in each genre

genre_counts = movies["genres"].str.split("|").explode().value_counts()
#explode() puts each genre into a separate row.
print("Number of movies in each genre:")
display(genre_counts)
```

```python
# Check the number of movies in each genre

genre_counts = movies["genres"].str.split("|").explode().value_counts()

print("Number of movies in each genre:")
display(genre_counts)
```

```python
# Check movies with no genre information

print("Movies with no genre:", (movies["genres"] == "(no genres listed)").sum())
```

```python
#Check Movie ID Consistency
# Check whether all rated movies exist in the movies dataset

movie_ids = set(movies["movieId"])
rating_movie_ids = set(ratings["movieId"])

missing_movie_ids = rating_movie_ids - movie_ids

print("Movie IDs missing from movies dataset:", len(missing_movie_ids))
```

```python
# Check timestamp values and data type

print(ratings["timestamp"].head())
print("Timestamp data type:", ratings["timestamp"].dtype)
```

```python
# EDA Summary


print("Unique movies:", movies["movieId"].nunique())
print("Unique users:", ratings["userId"].nunique())

print("Average rating:", ratings["rating"].mean())
print("Minimum rating:", ratings["rating"].min())
print("Maximum rating:", ratings["rating"].max())
```

## EDA Conclusion

- The MovieLens dataset contains movie details and user ratings.
- The movies dataset contains 9,742 records, while the ratings dataset contains 100,836 records.
- The `movieId` column uniquely identifies each movie in the movies dataset.
- The combination of `userId` and `movieId` uniquely identifies each rating record in the ratings dataset. This is called a **Composite Key**.
- The ratings dataset contains ratings given by different users to different movies.
- Genres are stored as multiple values separated by `|`.
- The `timestamp` column is stored as Unix time (integer format).
- All movie IDs in the ratings dataset have matching records in the movies dataset.
- Missing values and duplicate records were checked during data exploration.

EDA helped me understand the structure, relationships, and quality of the source datasets before starting the ETL process.
